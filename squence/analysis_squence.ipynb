```python
import pandas as pd
import matplotlib.pyplot as plt

# Membaca dataset
df = pd.read_csv("assets/AirQualityUCI.csv")
print("Dataset berhasil dimuat.")

# Menghapus kolom kosong
df = df.dropna(axis=1, how="all")

# Menggabungkan tanggal dan waktu menjadi sequence waktu
df["Datetime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"].str.replace(".", ":", regex=False),
    dayfirst=True,
    errors="coerce"
)

# Mengubah nilai -200 menjadi NaN
df = df.replace(-200, pd.NA)

# Mengurutkan berdasarkan waktu
df = df.sort_values("Datetime")

print(df.head())
print(df.shape)

# Mengecek data kosong
print(df.isna().sum())
print(df.describe())
```

```python
# Temperature sebagai squence
temperature = df[["Datetime", "T"]].dropna()

# Menampilkan sequence temperature
plt.figure(figsize=(12, 5))
plt.plot(temperature["Datetime"], temperature["T"])
plt.xlabel(("Waktu"))
plt.ylabel(("Temperatur (°C)"))
plt.title("Sequence Temperatur terhadap Waktu")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
